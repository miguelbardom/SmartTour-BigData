#### 1. ¿Por qué es importante utilizar Docker en proyectos Big Data?
Docker permite crear entornos aislados y reproducibles. De esta forma, todos los miembros del equipo pueden utilizar las mismas herramientas, versiones y configuraciones sin tener que instalar manualmente cada componente en sus ordenadores.

#### 2. ¿Qué ventajas tiene utilizar un contenedor frente a instalar todas las herramientas manualmente?
Un contenedor facilita la instalación y configuración del entorno, evita conflictos entre versiones y permite que el proyecto se ejecute de forma similar en diferentes ordenadores. Además, los contenedores se pueden crear, eliminar y reproducir fácilmente.

#### 3. ¿Por qué es importante organizar correctamente las carpetas de un proyecto?
Una estructura organizada facilita encontrar los archivos, mantener el proyecto y trabajar en equipo. También permite separar los datos, programas, notebooks y documentación, haciendo que el proyecto sea más fácil de mantener.

#### 4. ¿Qué diferencia existe entre una imagen Docker y un contenedor?
Una imagen Docker es una plantilla que contiene todo lo necesario para crear un entorno, como programas, librerías y configuración. Un contenedor es una instancia en ejecución creada a partir de esa imagen.
